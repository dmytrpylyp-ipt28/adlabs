Data Cleaning

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('diabetes_binary_health_indicators_BRFSS2015.csv')

print("Пропущені значення:")
print(df.isnull().sum())

print("\nТипи даних:")
print(df.dtypes)
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df.sample(1000), x='Age', y='BMI', hue='Diabetes_binary')
plt.title("Залежність BMI від вікової категорії")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='Diabetes_binary', y='BMI', data=df)
plt.title("Розподіл BMI для груп з діабетом та без")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x='Sex', hue='Smoker', data=df)
plt.title("Співвідношення курців за статтю (0 = жінка, 1 = чоловік)")
plt.legend(title='Smoker', labels=['Не курить', 'Курить'])
plt.show()
```

п.5

```python
plt.figure(figsize=(8,5))
plt.hist(df['GenHlth'], bins=5, color='skyblue', edgecolor='black')
plt.title("Розподіл загального стану здоров'я (5 категорій)")
plt.xlabel("Оцінка здоров'я")
plt.ylabel("Кількість людей")
plt.show()
```

```python
plt.figure(figsize=(12,10))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm')
plt.title("Матриця кореляції показників здоров'я")
plt.show()
```
