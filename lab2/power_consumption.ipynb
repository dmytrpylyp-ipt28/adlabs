Підготовка даних

```python
import pandas as pd
import timeit

df_p = pd.read_csv('household_power_consumption.txt', sep=';', low_memory=False)

df_p.replace('?', pd.NA, inplace=True)
df_p.dropna(inplace=True)

cols = ['Global_active_power', 'Global_reactive_power', 'Voltage', 
        'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']
df_p[cols] = df_p[cols].astype(float)

print("Дані підготовлено.")
```

Активна потужність > 5 кВт

```python
def task_1():
    return df_p[df_p['Global_active_power'] > 5.0]

t1 = timeit.timeit(task_1, number=1)
print(f"Результат (перші 5 рядків):\n{task_1().head()}")
print(f"Час виконання: {t1:.5f} сек")
```

Напруга та споживання (Струм 19-20 А)

```python
def task_2():
    return df_p[(df_p['Global_intensity'] >= 19) & (df_p['Global_intensity'] <= 20) & 
                (df_p['Sub_metering_2'] > df_p['Sub_metering_3'])]

t2 = timeit.timeit(task_2, number=1)
print(f"Знайдено записів: {len(task_2())}")
print(f"Час виконання: {t2:.5f} сек")
```

Випадкова вибірка 500,000 записів

```python
def task_3():
    sample = df_p.sample(n=500000, replace=False)
    return sample[['Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']].mean()

t3 = timeit.timeit(task_3, number=1)
print(f"Середні значення груп:\n{task_3()}")
print(f"Час виконання: {t3:.5f} сек")
```

Коефіцієнт Пірсона та Спірмена

```python
pearson = df_p['Global_active_power'].corr(df_p['Global_intensity'], method='pearson')
spearman = df_p['Global_active_power'].corr(df_p['Global_intensity'], method='spearman')

print(f"Коефіцієнт Пірсона: {pearson:.4f}")
print(f"Коефіцієнт Спірмена: {spearman:.4f}")
```
